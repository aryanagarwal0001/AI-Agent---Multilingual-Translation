```python
from langchain_classic.agents import AgentType, Tool, initialize_agent
from langchain_google_genai import ChatGoogleGenerativeAI
```

```python
llm = ChatGoogleGenerativeAI(model = "gemini-3.1-flash-lite", google_api_key = "YOUR_GOOGLE_API_KEY", temperature= 0)
```

```python
def translate_to_french(text):
    return f"Translate this text into French: {text}"

def translate_to_spainish(text):
    return f"Translate this text into Spainish: {text}"
```

```python
tools = [
    Tool(name = "French Translator", func= translate_to_french, description= "Translates text into french."),
    Tool(name = "Spainish Translator", func = translate_to_spainish, description = "Translates text into spainish")
]
```

```python
agent = initialize_agent(
    llm = llm,
    tools = tools,
    agent = AgentType.ZERO_SHOT_REACT_DESCRIPTION,
    verbose = True
)
```

```python
query = "I love learning AI."

response = agent.run(query)
print(response)
```
